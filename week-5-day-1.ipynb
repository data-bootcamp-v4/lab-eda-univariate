```python
#Business Question: How are products priced on Amazon UK, and are there specific price points or ranges that are more common?

#Measures of Centrality:

#Calculate the mean, median, and mode for the price of products.

mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

# Print mean, median, and mode
print("Mean Price:", mean_price)
print("Median Price:", median_price)
print("Mode Price:", mode_price)

#What's the average price point of products listed? How does this compare with the most common price point (mode)?
```

```python
#Measures of Dispersion:

#Determine the variance, standard deviation, range, and interquartile range for product price.

variance = df['price'].var()
std_dev = df['price'].std()
print("varrrr:", variance)
print("std dev:", std_dev)

price_range = df['price'].max() - df['price'].min()
price_iqr = df['price'].quantile(0.75) - df['price'].quantile(0.25)
print("Range of product price:", price_range)
print("Interquartile range of product price:", price_iqr)


#How varied are the product prices? Are there any indicators of a significant spread in prices?
```

```python
#Visualizations:

#Is there a specific price range where most products fall? Plot a histogram to visualize the 
#distribution of product prices. If its hard to read these diagrams, think why this is, 
#and explain how it could be solved..

plt.figure(figsize=(10, 6))
sns.histplot(df['price'], bins=np.linspace(df['price'].min(), df['price'].max(), num=20), kde=True)
plt.title('Histogram of Product Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

#Are there products that are priced significantly higher than the rest? Use a box plot to 
#showcase the spread and potential outliers in product pricing.

plt.figure(figsize=(10, 6))
sns.boxplot(x=df['price'])
plt.title('Box Plot of Prices')
plt.xlabel('prices')
plt.grid(True)
plt.show()
```

```python
# Measures of Dispersion
rating_variance = df['stars'].var()
rating_std_dev = df['stars'].std()
rating_iqr = df['stars'].quantile(0.75) - df['stars'].quantile(0.25)

print("Variance of product rating:", rating_variance)
print("Standard deviation of product rating:", rating_std_dev)
print("Interquartile range of product rating:", rating_iqr)

# Shape of the Distribution
rating_skewness = df['stars'].skew()
rating_kurtosis = df['stars'].kurtosis()

print("Skewness of product rating:", rating_skewness)
print("Kurtosis of product rating:", rating_kurtosis)

# Visualizations
plt.figure(figsize=(10, 6))
sns.histplot(df['stars'], bins=20, kde=True)
plt.title('Histogram of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```
