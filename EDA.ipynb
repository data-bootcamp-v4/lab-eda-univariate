#### **Objective**: Explore the product listing dynamics on Amazon UK to extract actionable business insights. By understanding the distribution, central tendencies, and relationships of various product attributes, businesses can make more informed decisions on product positioning, pricing strategies, and inventory management.

```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import numpy as np

```

## import the data

```python
df = pd.read_csv('amz_uk_price_prediction_dataset.csv')
```

```python
display(df.head(2))
df.shape
```

## Understand column: category

#### the most popular product categories on Amazon UK is Sports & Outdoors, Beauty	, Handmade Clothing, Shoes & Accessories ....

```python
df['category'].dtypes
df['category'].isna().sum()
```

```python
df_category_frequency = df.groupby('category')['uid'].agg('count').reset_index(name='listing_count')
df_category_frequency.sort_values(by='listing_count',ascending= False,inplace=True)
# df_category_frequency['listing_proportion'] = df_category_frequency['listing_count']/df_category_frequency['listing_count'].sum()

# a frequency table for the product `category`
print(df_category_frequency)

# top 5 categories 
df_category_frequency_top5 = df_category_frequency.head(5).reset_index(drop=True)
df_category_frequency_top5
```

#### bar plot to display listing frequency for top 5 categories
#### visualize proportions for top 5 categories using a pie chart.

```python
# bar chart to visualize count of listing for top 5 categories
plt.figure(figsize=(20,5))
plt.bar(df_category_frequency_top5['category'],df_category_frequency_top5['listing_count'],color='blue')
plt.xlabel('Category')
plt.ylabel('Count of listing')
plt.title('Top 5 Category Listing Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# pie chart to visualize the proportion of listing for top 5 categories
# TAKEAWAY--> sports and outdoors category dominate the listings

plt.figure(figsize=(8,8))
plt.pie(
    df_category_frequency_top5['listing_count'],
    labels=None,  
    autopct='%1.1f%%',  # show percentages
    startangle=90,
    colors=plt.cm.tab20.colors
)

plt.title('Category Proportion')
plt.axis('equal')

# Add a legend outside the pie
plt.legend(
    df_category_frequency_top5['category'],
    title="Category",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1)  # move legend outside
)

plt.tight_layout()
plt.show()
```

## Understand column: Product price

```python
df['price'].dtype
df['price'].isna().sum()
```

### Centrality

##### the average price is much higher than the mode , 
##### there are a few expensive products that are pulling up the average price 
##### it tells us the averge price seems expensive if you look at the mean, but actually most products are priced close to 9.99

```python
price_mean = df['price'].mean()
print(f"price mean is :{price_mean:.2f}")

price_median = df['price'].median()
print(f"price median is :{price_median:.2f}")

df['price'].mode() # Series
price_mode = df['price'].mode().iloc[0]
print(f"price mode is :{price_mode:.2f}")

```

### Dispersion

##### product prices are highly varied
##### the combination of large standard deviation, huge range, tight IRQ tells that most products are cheap, but a few expensive ones creates a significant spread

```python
price_variance = df['price'].var()
print(f"price variance is :{price_variance:.2f}")

price_std = df['price'].std()
print(f"price standard deviation is :{price_std:.2f}")

price_range = df['price'].max() - df['price'].min()
print(f"price range is :{price_range:.2f}")

price_Q3 = df['price'].quantile(0.75)
price_Q1 = df['price'].quantile(0.25)
price_IQR = price_Q3 - price_Q1
print(f"middle 50% of price are between {price_Q1} and {price_Q3}")

# How varied are the product prices? Are there any indicators of a significant spread in prices?
```

### visualization

#### based on histogram, most products fall in price range (9,10)
#### based on boxplot, two products are priced significantly higher than the rest 

```python
# histogram --> distribution of product price 
plt.figure(figsize=(30,8))

n_bin = 101
bin_edges = np.linspace(0,100,n_bin)

plt.hist(df['price'], bins=bin_edges, color='skyblue', edgecolor='black')
plt.title('Distribution of Product Prices')
plt.xlabel('Price')
plt.ylabel('Listing_Count')

plt.xticks(bin_edges, labels=[f'{int(p)}' for p in bin_edges], rotation=45)
plt.show()


```

```python
# boxplot --- distribution of product price

plt.figure(figsize=(30,4))

plt.boxplot(df['price'], vert=False, showfliers=True)
plt.title("Boxplot of Product Prices")
plt.xlabel("Price")
plt.show()
```

## Understand column: Product Ratings

```python
df['stars'].dtype
df['stars'].isna().sum()
```

### Centrality
#### at least 50% listings have rating 0.0/no rating, the rating appears most frequently is 0.It tells us most customers generally do not provide ratings.

```python
rating_mean = df['stars'].mean()
print(f"rating mean is :{rating_mean}")

rating_median = df['stars'].median()
print(f"rating median is :{rating_median}")

df['stars'].mode() # Series
rating_mode = df['stars'].mode().iloc[0]
print(f"rating mode is :{rating_mode}")
```

### Dispersion

#### ratings are not consistent,SD is high relative to rating scale.
#### Customer feedback is all over the place rather than clustered around a typical score.

```python
rating_variance = df['stars'].var()
print(f"rating variance is :{rating_variance:.2f}")

rating_std = df['stars'].std()
print(f"rating standard deviation is :{rating_std:.2f}")

rating_range = df['stars'].max() - df['stars'].min()
print(f"rating range is :{rating_range:.2f}")

rating_Q3 = df['stars'].quantile(0.75)
rating_Q1 = df['stars'].quantile(0.25)
rating_IQR = rating_Q3 - rating_Q1
print(f"middle 50% of rating are between {rating_Q1} and {rating_Q3}")
```

### shape of distribution: skewness and kurtosis

```python
# skewness >0, distribution is right-skewed (tail on the right), mean>median
# but close to 0 , distribution is roughly symmetric
df['stars'].skew()
```

```python
# tailedness <0, light tails
# fewer extreme ratings than normal distribution
df['stars'].kurt()
```

### visualization of distribution

```python
plt.figure(figsize=(10,5))

n_bin_rating = 6 # 6 points
bin_edges_rating = np.linspace(0,5,n_bin_rating) # 5 bins
plt.hist(df['stars'], bins=bin_edges_rating, color='skyblue', edgecolor='black')
plt.title('Distribution of Product Rating')
plt.xlabel('rating')
plt.ylabel('frequency')

plt.xticks(bin_edges_rating, labels=[f'{int(p)}' for p in bin_edges_rating], rotation=45)
plt.show()
```
