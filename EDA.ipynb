```python
Medidas de centralidad :
```

```python
# Calcular la media, la mediana y la moda para los price productos.
price_mean = data_amazon['price'].mean()
price_median = data_amazon['price'].median()
price_mode = data_amazon['price'].mode().iloc[0]

print("Media del precio de los productos:", price_mean)
print("Mediana del precio de los productos:", price_median)
print("Moda del precio de los productos:", price_mode)
```

```python
# ¿Cuál es el precio promedio de los productos que se muestran?

average_price = data_amazon ['price'].mean()
print (f"El precio promedio de los productos es : {average_price}")
```

```python
# ¿Cómo se compara con el precio más común (moda)?

mode_price = data_amazon ['price'].mode()
print (f"El precio más común es : {mode_price}")
```

```python
print(f"\nLa media del precio es {price_mean:.2f} y la moda del precio es {price_mode:.2f}.")
print("¿Cómo se comparan estos dos valores")
if price_mean > price_mode :
   print("La media es mayor que la moda, lo que sugiere que la distribución puede estar sesgada hacia los precios más altos.")
elif price_mean < price_mode:
    print("La moda es mayor que la media, lo que sugiere que la mayoría de los precios están cerca del valor más bajo.")
else:
    print("La media y la moda son iguales, lo que sugiere que los precios pueden estar distribuidos uniformemente.")
```

```python
Medidas de dispersión :
```

```python
variance_price = data_amazon['price'].var()

std_dev_price = data_amazon['price'].std()

range_price = data_amazon['price'].max() - data_amazon['price'].min()

iqr_price = data_amazon['price'].quantile(0.75) - data_amazon['price'].quantile(0.25)

print("Variance of Price:", variance_price)
print("Standard Deviation of Price:", std_dev_price)
print("Range of Price:", range_price)
print("Interquartile Range of Price:", iqr_price)
```

```python
# Gráfico de barras
plt.figure(figsize=(16, 8))  
sns.histplot(data_amazon['price'], bins=20, kde=True, color='blue', stat='density')

plt.title('Distribución de los Precios de los Productos')
plt.xlabel('Precio')
plt.ylabel('Densidad')

plt.show()
```

```python
# Diagrama de cajas
plt.figure(figsize=(16, 8))  

# Crear el boxplot
sns.boxplot(x=data_amazon['price'], color='green')
plt.title('Diagrama de Cajas de los Precios de los Productos', fontsize=16)
plt.xlabel('Precio', fontsize=14)
plt.xticks(fontsize=12)  
plt.xticks(rotation=45, ha='right') 

plt.show()
```

Calcular la media, la mediana y la moda para los rating productos.
¿Cómo califican los clientes los productos en general? ¿Existe una tendencia común?
Medidas de dispersión :

Determine la varianza, la desviación estándar y el rango intercuartil para el producto rating.
¿Son consistentes las calificaciones o hay una amplia variación en los comentarios de los clientes?
Forma de la distribución :

Calcular la asimetría y la curtosis de la ratingcolumna.
¿Las calificaciones se distribuyen normalmente o tienden hacia valores más altos o más bajos?

```python
median_ratings = data_amazon['stars'].median()
mode_ratings = data_amazon['stars'].mode().iloc[0]
mean_ratings = data_amazon['stars'].mean()

print(f"La media de las calificaciones es {mean_ratings:.2f}, la mediana es {median_ratings:.2f} y la moda es {mode_ratings:.2f}.")

print(median_ratings)
print(mode_ratings)
print(mean_ratings)
```

```python
variance_ratings = data_amazon['stars'].var()

std_dev_ratings = data_amazon['stars'].std()

range_ratings = data_amazon['stars'].max() - data_amazon['stars'].min()

iqr_ratings = data_amazon['stars'].quantile(0.75) - data_amazon['stars'].quantile(0.25)

print("Variance of Ratings:", variance_ratings)
print("Standard Deviation of Ratings:", std_dev_ratings)
print("Range of Ratings:", range_ratings)
print("Interquartile Range of Ratings:", iqr_ratings)
```

```python
skewness_rating = data_amazon['stars'].skew()

print(f"Asimetría de las calificaciones: {skewness_rating}")
```

```python
kurtosis_rating = data_amazon['stars'].kurt()

print(f"Curtosis de las calificaciones: {kurtosis_rating}")
```

```python
plt.figure(figsize=(16, 8))
sns.histplot(data_amazon['stars'], bins=20, kde=True, color='blue', stat='density')

plt.title('Distribución de las Calificaciones de los Productos')
plt.xlabel('Calificaciones')
plt.ylabel('Densidad')

plt.show()
```
