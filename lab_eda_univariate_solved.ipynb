```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "/Users/borjacervera/Documents/Ironhack/Labs/Unit5_W13/amz_uk_price_prediction_dataset.csv"
df = pd.read_csv(csv_path)
```

```python
df.head()
```

```python
df.columns
```

```python
category_column = "category"
category_counts = df[category_column].value_counts()
print(category_counts)
```

```python
top_categories = category_counts.head(5)
print(top_categories)
```

```python
plt.figure(figsize=(10, 6))
top_categories.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Top 5 Categorías Más Comunes en Amazon UK")
plt.xlabel("Categoría")
plt.ylabel("Frecuencia")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 8))
top_categories.plot(kind="pie", autopct="%1.1f%%", colors=["skyblue", "lightcoral", "gold", "lightgreen", "orange"])
plt.title("Distribución de las 5 Categorías Más Comunes")
plt.ylabel("")  # Ocultar etiqueta del eje Y
plt.show()
```

```python
price_column = "price"  

mean_price = df[price_column].mean()
median_price = df[price_column].median()
mode_price = df[price_column].mode()[0]

print(f"Mean price: {mean_price}")
print(f"Median price: {median_price}")
print(f"Mode price: {mode_price}")
```

```python
# Calcular medidas de dispersión
variance_price = df[price_column].var()
std_dev_price = df[price_column].std()
price_range = df[price_column].max() - df[price_column].min()
iqr_price = df[price_column].quantile(0.75) - df[price_column].quantile(0.25)

print(f"Variance: {variance_price}")
print(f"Standard deviation: {std_dev_price}")
print(f"Price range: {price_range}")
print(f"IQR: {iqr_price}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[price_column], bins=50, kde=True)
plt.xlabel("Precio")
plt.ylabel("Frecuencia")
plt.title("Distribución de Precios de Productos en Amazon UK")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df[price_column])
plt.xlabel("Precio")
plt.title("Boxplot de Precios de Productos en Amazon UK")
plt.show()
```

```python
rating_column = "reviews"  

mean_rating = df[rating_column].mean()
median_rating = df[rating_column].median()
mode_rating = df[rating_column].mode()[0]

print(f"Rating mean: {mean_rating}")
print(f"Rating median: {median_rating}")
print(f"Rating mode: {mode_rating}")
```

```python
variance_rating = df[rating_column].var()
std_dev_rating = df[rating_column].std()
iqr_rating = df[rating_column].quantile(0.75) - df[rating_column].quantile(0.25)

print(f"Variance: {variance_rating}")
print(f"Standard deviation: {std_dev_rating}")
print(f"IQR: {iqr_rating}")
```

```python
from scipy.stats import skew, kurtosis
```

```python
skewness = skew(df[rating_column].dropna()) 
kurt = kurtosis(df[rating_column].dropna())

print(f"Skewness: {skewness}")
print(f"Curtosis: {kurt}")
```
