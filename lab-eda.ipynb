```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%matplotlib inline
```

```python
df = pd.read_csv('amz_uk_price_prediction_dataset.csv')
```

```python
df
```

```python
df.columns
```

```python
#part 1
#Frequency Tables
# Frequency tables. Counts and proportions. # Visualizations: Bar charts, pie charts
#the frequency of each category

category_counts = df['category'].value_counts()
print(category_counts.head(5))
```

```python
top_5_categories = df['category'].value_counts().head(5)
print(top_5_categories)
```

```python
#2 Visualizations

#Bar Chart for Distribution of Products in Categories

# Plotting the bar chart
sns.barplot(x=top_5_categories.index, y=top_5_categories.values, palette="Set3")
```

```python
#Pie chart




# Plotting the pie chart
plt.figure(figsize=(8, 8))
plt.pie(top_5_categories, labels=top_5_categories.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette("Set3"))
plt.title('Proportion of Top 5 Product Categories')
```

```python
df.describe()
```

```python
#Part 2 Delving into Product Pricing

#1 Measures of Centrality:

mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

mean_price, median_price, mode_price
```

```python
# Measures of Dispersion:

variance_price = df['price'].var()
std_dev_price = df['price'].std()
min_price = df['price'].min()
max_price = df['price'].max()
range_price = max_price - min_price
quantiles_price = df['price'].quantile([0.25, 0.5, 0.75])

variance_price, std_dev_price, min_price, max_price, range_price, quantiles_price
```

```python
df['price'].quantile(0.2)
```

```python
skewness_price = df['price'].skew()
kurtosis_price = df['price'].kurtosis()

skewness_price, kurtosis_price
```

```python
# Visualizations:

plt.figure(figsize=(10, 6))  # Set the size of the plot
sns.histplot(df['price'], kde=True, bins=30, color="salmon")

```

```python
#box plot product price significantly higher than the rest?
plt.title('Product Prices')  
plt.xlabel('Price') 
sns.boxplot(data = df['price'], color="blue")
```

```python
#Part 3: Unpacking Product Ratings

mean_price = df['stars'].mean()
median_price = df['stars'].median()
mode_price = df['stars'].mode()[0]

mean_price, median_price, mode_price
```

```python
#Dispersion:

# Measures of Dispersion:

variance_price = df['stars'].var()
std_dev_price = df['stars'].std()
min_price = df['stars'].min()
max_price = df['stars'].max()
range_price = max_price - min_price
quantiles_price = df['stars'].quantile([0.25, 0.5, 0.75])

variance_price, std_dev_price, min_price, max_price, range_price, quantiles_price
```

```python
skewness_price = df['stars'].skew()
kurtosis_price = df['stars'].kurtosis()

skewness_price, kurtosis_price
```

```python
# Creating the histogram
plt.figure(figsize=(10, 6)) 
sns.histplot(df['stars'], bins=5, kde=False, color='skyblue')  
plt.title('Distribution of Product Ratings')  # Adding a title
plt.xlabel('Rating (Stars)')  
plt.ylabel('Frequency')  # 
plt.xticks(range(1, 6))  #  ratings are from 1 to 5
plt.show()
```
