```python
top_categories = df['category'].value_counts().head(10)
top_categories
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
top_categories.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Top 10 Categories by Product Count')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    top_categories,
    labels=top_categories.index,
    autopct='%1.1f%%',    # show percentages with 1 decimal
    startangle=90,        # start from the top
    counterclock=False    # clockwise order
)

plt.title('Top 10 Categories - Product Distribution')
plt.show()
```

```python
df['price'].mode()
df['price'].mean()
```

```python
plt.figure(figsize=(10,6))
plt.hist(df['price'], bins=30, color='skyblue', edgecolor='black')  # 30 bins
plt.title('Histogram of Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10,6))
sns.histplot(df['price'], bins=10, kde=False, color='skyblue')
plt.title('Histogram of Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.boxplot(df['price'], vert=True, patch_artist=True, 
            boxprops=dict(facecolor='skyblue', color='black'),
            medianprops=dict(color='red'),
            whiskerprops=dict(color='black'),
            capprops=dict(color='black'))
plt.title('Box Plot of Price')
plt.ylabel('Price')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
display(df['stars'].mode())
df['stars'].mean()
df['stars'].median()
df['stars'].describe()
```

```python
round(df['stars'].var(), 2)
```

```python
# Calculate skewness
skewness = df['stars'].skew()
print("Skewness:", round(skewness, 2))

# Calculate kurtosis
kurtosis = df['stars'].kurt()
print("Kurtosis:", round(kurtosis, 2))
```

```python
plt.hist(df['stars'], bins=30, color='skyblue', edgecolor='black')  # 30 bins
plt.title('Histogram of Stars')
plt.xlabel('Stars')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
