```python
mean_price =   df["price"].mean()
median_price = df["price"].median()
mode_price =    df["price"].mode()[0]
```

```python
mean_price, median_price, mode_price
```

**Insights** 

Average Price vs. Most Common Price Point:
Average Price Point (Mean): The average price of products listed on Amazon UK is £89.24.
Most Common Price Point (Mode): The most common price point for products is £9.99.

The average price (£89.24) is nearly 9 times higher than the most common price point (£9.99), and significantly higher than the median price (£19.09).
This substantial gap indicates a strong right-skewed (positively skewed) distribution.

```python
variance_price = df["price"].var()
std_dev_price  = df["price"].std()
min_price      = df["price"].min()
max_price      = df["price"].max()
range_price    = max_price - min_price
quantiles_price = df["price"].quantile([0.25, 0.5, 0.75])

variance_price, std_dev_price, min_price, max_price, range_price, quantiles_price
```

```python
df['price'].quantile(0.1)
```

```python
sns.histplot(df['price'], kde=True, bins=30, color="salmon")
```

```python
p95 = df["price"].quantile(0.95)

plt.figure(figsize=(8, 5))

sns.histplot(df[df["price"] <= p95]["price"], bins=30, kde=True, color="salmon")

```

```python
sns.boxplot(data = df['price'], color="lightblue")
```

**Insights** 

The raw dataset contains extreme price outliers (e.g., listings priced up to £100,000). These extreme values stretched the X-axis dramatically, compressing 95% of the data into a single, unreadable bar near £0.
By filtering out extreme outliers using the 95th percentile threshold (price <= p95), we effectively removed the long tail of high-value anomalies while preserving 95% of the platform's standard product listings (prices between £0 and ~£350).

```python
ratings = df["stars"].dropna()
```

```python
rating_mean = ratings.mean()
rating_median = ratings.median()
rating_mode = ratings.mode()[0]

rating_mean, rating_median, rating_mode
```

```python
rating_var = ratings.var()
rating_std = ratings.std()
quantiles_price = df["stars"].quantile([0.25, 0.5, 0.75])
price_range = df["stars"].max() - df["stars"].min()

rating_var, rating_std, quantiles_price, price_range
```

```python
rating_skew = ratings.skew()
rating_kurto = ratings.kurtosis()

rating_skew, rating_kurto
```

```python
sns.histplot(df['stars'], kde=True, bins=30, color="salmon")
```

**Insights** 

1. Measures of Centrality & Common Trend
Findings:

Mean: 2.15

Median: 0.0

Mode: 0.0

Analysis & Trend: The overall rating statistics are heavily influenced by a massive proportion of products with a 0.0 rating. Rather than indicating customer dissatisfaction, a 0.0 rating represents unrated products (products with 0 customer reviews) on Amazon UK.

2. Measures of Dispersion & Feedback ConsistencyFindings:Variance: 4.82Standard Deviation: 2.19Quantiles: Q_1 (25\%) = 0.0, $Q_2 (50\%) = 0.0, Q_3 (75\%) = 4.4 IQR: 4.4 - 0.0 = 4.4 Total Range: 5.0 (0.0 to 5.0)Analysis: There is a wide variation in the dataset (Std Dev = 2.19). More than 50% of all listed products have no reviews at all (0.0 star), whereas rated products quickly jump into the high-rating territory (with Q_3 reaching 4.4 stars).

3. Shape of the Distribution
Findings:

Skewness: 0.0812 (slight right skew overall, but highly bimodal)

Kurtosis: -1.9260 (Platykurtic / multi-peaked)

Analysis: The ratings are not normally distributed. The negative kurtosis confirms a flatter, dual-clustered distribution. The distribution is strongly bimodal, exhibiting two distinct groups:

A dominant spike at 0.0 (Unrated items).

A secondary, healthy curve centered between 4.0 and 4.8 stars (Rated items).

4. Visualizations & Most Common Rating
Histogram Observations: The histogram clearly illustrates a bimodal (two-peak) distribution:

Peak 1 (Highest Frequency): 0.0 stars , representing unrated items.

Peak 2 (Most Common Rated Value): Among products that do have customer reviews, the most common rating point sits around 4.5 stars.

Business Takeaway: A vast proportion of listings on Amazon UK suffer from zero customer engagement. However, once a product receives reviews, customer satisfaction is generally high, with ratings strongly clustering between 4.0 and 4.6 stars.
