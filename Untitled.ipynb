```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Log-transform the data
df['Logprice'] = np.log1p(df['price'])  # log(1 + x) avoids log(0) issues

# Plot
plt.figure(figsize=(8, 5))
sns.histplot(df['Logprice'], bins=30, kde=True)
plt.title('Distribution of Log-Transformed GrLivArea')
plt.xlabel('Log(price + 1)')
plt.tight_layout()
plt.show()
```
