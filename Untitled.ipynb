```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os
%matplotlib inline
```

```python
# Path to the downloaded file
file_path = 'amz_uk_price_prediction_dataset.csv'

# Read the dataset
df = pd.read_csv(file_path)
df.head()
```

```python
# Generate frequency table for product category
category_freq = df['category'].value_counts()

# Display the frequency table
print("Frequency Table for Product Categories:")
print(category_freq)

# Identify the top 5 most listed categories
top_5_categories = category_freq.head(5)
print("\nTop 5 Most Listed Product Categories:")
print(top_5_categories)
```

```python
# Bar chart for all categories (or top N if too many categories)
plt.figure(figsize=(12, 6))
# Limit to top 10 categories for readability
top_10_categories = category_freq.head(10)
sns.barplot(x=top_10_categories.values, y=top_10_categories.index)
plt.title('Distribution of Products Across Top 10 Categories')
plt.xlabel('Number of Listings')
plt.ylabel('Category')
plt.tight_layout()
plt.show()

# Pie chart for top 5 categories
plt.figure(figsize=(8, 8))
plt.pie(top_5_categories, labels=top_5_categories.index, autopct='%1.1f%%', startangle=140)
plt.title('Proportion of Top 5 Product Categories')
plt.show()
```

```python
# Calculate measures of centrality for price
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]  # Take the first mode if multiple

# Display results
print("Measures of Centrality for Price:")
print(f"Mean Price: £{mean_price:.2f}")
print(f"Median Price: £{median_price:.2f}")
print(f"Mode Price: £{mode_price:.2f}")

# Comparison
print("\nComparison:")
print(f"The average price (£{mean_price:.2f}) is {'higher' if mean_price > mode_price else 'lower'} than the most common price (£{mode_price:.2f}).")
```

```python
# Calculate measures of dispersion for price
variance_price = df['price'].var()
std_price = df['price'].std()
range_price = df['price'].max() - df['price'].min()
iqr_price = df['price'].quantile(0.75) - df['price'].quantile(0.25)

# Display results
print("Measures of Dispersion for Price:")
print(f"Variance: {variance_price:.2f}")
print(f"Standard Deviation: £{std_price:.2f}")
print(f"Range: £{range_price:.2f}")
print(f"Interquartile Range: £{iqr_price:.2f}")
```

```python
# Histogram for price distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['price'], bins=50, kde=True)
plt.title('Distribution of Product Prices')
plt.xlabel('Price (£)')
plt.ylabel('Frequency')
plt.show()

# Box plot for price spread and outliers
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['price'])
plt.title('Box Plot of Product Prices')
plt.ylabel('Price (£)')
plt.show()
```

```python
# Calculate measures of centrality for rating
mean_rating = df['stars'].mean()
median_rating = df['stars'].median()
mode_rating = df['stars'].mode()[0]

# Display results
print("Measures of Centrality for stars:")
print(f"Mean Rating: {mean_rating:.2f}")
print(f"Median Rating: {median_rating:.2f}")
print(f"Mode Rating: {mode_rating:.2f}")
```

```python
# Calculate measures of dispersion for stars
variance_stars = df['stars'].var()
std_stars = df['stars'].std()
iqr_stars = df['stars'].quantile(0.75) - df['stars'].quantile(0.25)

# Display results
print("Measures of Dispersion for Stars:")
print(f"Variance: {variance_stars:.2f}")
print(f"Standard Deviation: {std_stars:.2f}")
print(f"Interquartile Range: {iqr_stars:.2f}")
```

```python
# Calculate skewness and kurtosis for rating
skewness_rating = df['stars'].skew()
kurtosis_rating = df['stars'].kurtosis()

# Display results
print("Shape of Rating Distribution:")
print(f"Skewness: {skewness_rating:.2f}")
print(f"Kurtosis: {kurtosis_rating:.2f}")
```

```python
# Histogram for rating distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['stars'], bins=20, kde=True)
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```
