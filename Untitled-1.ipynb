```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# carregar dataset
df = pd.read_csv("amazon_uk_product_data.csv")

df.head()
df.info()
```

```python
category_freq = df["category"].value_counts()

category_freq.head(10)
```

```python
top5_categories = category_freq.head(5)
top5_categories
```

```python
top10_categories = category_freq.head(10)

plt.figure()
top10_categories.plot(kind="bar")
plt.xticks(rotation=45)
plt.title("Top 10 Product Categories")
plt.ylabel("Number of Listings")
plt.show()
```

```python
plt.figure()
top5_categories.plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Top 5 Category Distribution")
plt.show()
```

```python
mean_price = df["price"].mean()
median_price = df["price"].median()
mode_price = df["price"].mode()[0]

mean_price, median_price, mode_price
```

```python
variance_price = df["price"].var()
std_price = df["price"].std()
price_range = df["price"].max() - df["price"].min()
iqr_price = df["price"].quantile(0.75) - df["price"].quantile(0.25)

variance_price, std_price, price_range, iqr_price
```

```python
plt.figure()
plt.hist(df["price"], bins=50)
plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()
plt.boxplot(df["price"], vert=False)
plt.title("Boxplot of Product Prices")
plt.show()
```

```python
mean_rating = df["rating"].mean()
median_rating = df["rating"].median()
mode_rating = df["rating"].mode()[0]

mean_rating, median_rating, mode_rating
```

```python
variance_rating = df["rating"].var()
std_rating = df["rating"].std()
iqr_rating = df["rating"].quantile(0.75) - df["rating"].quantile(0.25)

variance_rating, std_rating, iqr_rating
```

```python
skew_rating = df["rating"].skew()
kurt_rating = df["rating"].kurt()

skew_rating, kurt_rating
```

```python
plt.figure()
plt.hist(df["rating"], bins=20)
plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```
