```python
import pandas as pd
df = pd.read_csv("amazon_uk_products.csv")   # adjust filename
df.head()
```

```python
cat_freq = df['category'].value_counts()
print(cat_freq.head(10))
```

```python
top5_categories = cat_freq.head(5)
print("Top 5 Most Listed Categories:")
print(top5_categories)
```

```python
import matplotlib.pyplot as plt

# Bar chart for top 10
cat_freq.head(10).plot(kind="barh", figsize=(8,5))
plt.xlabel("Number of Listings")
plt.ylabel("Category")
plt.title("Top 10 Product Categories on Amazon UK")
plt.gca().invert_yaxis()
plt.show()
```

```python
top5_categories.plot(kind="pie", autopct="%1.1f%%", figsize=(6,6))
plt.title("Proportion of Listings – Top 5 Categories")
plt.ylabel("")
plt.show()
```

```python
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

print("Mean:", mean_price)
print("Median:", median_price)
print("Mode:", mode_price)
```

```python
variance = df['price'].var()
std_dev = df['price'].std()
price_range = df['price'].max() - df['price'].min()
iqr = df['price'].quantile(0.75) - df['price'].quantile(0.25)

print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Range:", price_range)
print("IQR:", iqr)
```

```python
plt.figure(figsize=(8,5))
df['price'].plot(kind="hist", bins=50, edgecolor="black")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of Product Prices")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# --- Step 1: Clean the stars column ---
# Extract numeric values from strings like "4.3 out of 5 stars"
df['stars'] = df['stars'].astype(str).str.extract(r"([0-9]+(?:\.[0-9]+)?)").astype(float)

# --- Step 2: Measures of Centrality ---
mean_rating = df['stars'].mean()
median_rating = df['stars'].median()
mode_rating = df['stars'].mode()[0]

print("Mean Rating:", mean_rating)
print("Median Rating:", median_rating)
print("Mode Rating:", mode_rating)

# --- Step 3: Measures of Dispersion ---
variance = df['stars'].var()
std_dev = df['stars'].std()
iqr = df['stars'].quantile(0.75) - df['stars'].quantile(0.25)

print("\nVariance:", variance)
print("Standard Deviation:", std_dev)
print("IQR (Interquartile Range):", iqr)

# --- Step 4: Shape of Distribution ---
skewness = skew(df['stars'].dropna())
kurt = kurtosis(df['stars'].dropna())

print("\nSkewness:", skewness)
print("Kurtosis:", kurt)

# --- Step 5: Visualization ---
plt.figure(figsize=(7,5))
df['stars'].plot(kind="hist", bins=10, edgecolor="black")
plt.xlabel("Rating (Stars)")
plt.ylabel("Frequency")
plt.title("Distribution of Product Ratings (Amazon UK)")
plt.show()
```
