```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('amz_uk_price_prediction_dataset.csv')
```

```python
df.head()
```

```python
# Frequency table for product categories
category_freq = df['category'].value_counts()
print(category_freq)
```

```python
# Display the top 5 categories
top_5_categories = category_freq.head(5)
print("Top 5 Categories:")
print(top_5_categories)
```

```python
#Sports & Outdoors is clearly the most requested category that dominates the listings
```

```python
plt.figure(figsize=(12, 6))
category_freq.plot(kind='bar', color='skyblue')
plt.title('Product Category Distribution')
plt.xlabel('Category')
plt.ylabel('Frequency')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
top_5_categories = category_freq.head(5)  # Select the top 5 categories
top_5_categories.plot(kind='bar', color='skyblue', label='Number of Listings')  # Add a label
plt.title('Top 5 Product Category Distribution')
plt.xlabel('Category')
plt.ylabel('Frequency')
plt.xticks(rotation=45)  # Slight rotation for better readability
plt.legend(loc='upper right')  # Add a legend
plt.show()
```

```python
plt.figure(figsize=(8, 8))
top_5_categories.plot(kind='pie', autopct='%1.1f%%', colors=sns.color_palette('pastel'))
plt.title('Proportion of Top 5 Product Categories')
plt.ylabel('')  # Hides the y-axis label
plt.show()
```

```python
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

print(f"Mean Price: {mean_price}")
print(f"Median Price: {median_price}")
print(f"Mode Price: {mode_price}")
```

```python
#The mean is 89euro , aprox , the median 19,09 and the price that is more recurrent is 9,99
```

```python
variance_price = df['price'].var()
std_dev_price = df['price'].std()
range_price = df['price'].max() - df['price'].min()
q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)
iqr_price = q3 - q1

print(f"Variance: {variance_price}")
print(f"Standard Deviation: {std_dev_price}")
print(f"Range: {range_price}")
print(f"Interquartile Range (IQR): {iqr_price}")
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(df['price'], bins=50, kde=True, color='purple')
plt.title('Distribution of Product Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['price'], color='orange')
plt.title('Box Plot of Product Prices')
plt.xlabel('Price')
plt.grid(True)
plt.show()
```

```python
mean_rating = df['stars'].mean()
median_rating = df['stars'].median()
mode_rating = df['stars'].mode()[0]

print(f"Mean Rating: {mean_rating}")
print(f"Median Rating: {median_rating}")
print(f"Mode Rating: {mode_rating}")
```

```python
#the rates are negative , but it is because most of the ratings is eqaul to 0
```

```python
variance_rating = df['stars'].var()
std_dev_rating = df['stars'].std()
q1_rating = df['stars'].quantile(0.25)
q3_rating = df['stars'].quantile(0.75)
iqr_rating = q3_rating - q1_rating

print(f"Variance: {variance_rating}")
print(f"Standard Deviation: {std_dev_rating}")
print(f"Interquartile Range (IQR): {iqr_rating}")
```

```python
skewness = df['stars'].skew()
kurtosis = df['stars'].kurt()

print(f"Skewness: {skewness}")
print(f"Kurtosis: {kurtosis}")
```
