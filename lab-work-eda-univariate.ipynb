```python
 # How are products priced on Amazon UK, and are there specific price points or ranges that are more common?
# the most frequent price is 9.99 but mean price is 89.24. Also after making var analysis we can see a very big spred it means tha there are some rare products with very big price.
# we can see it most ckear on boxplot.



# Measures of Centrality**:
    # Calculate the mean, median, and mode for the `price` of products.
    # What's the average price point of products listed? = 89.24380943923663 How does this compare with the most common price point (mode)? = 9.99 
```

```python
df.describe()
```

```python
mode = df['price'].mode()[0]
med = df['price'].median()
mean = df['price'].mean()

print(mode)
print(med)
print(mean)
```

```python
# Measures of Dispersion**:
    # Determine the variance, standard deviation, range, and interquartile range for product `price`.
    # How varied are the product prices? Are there any indicators of a significant spread in prices? 
# Product price is very varied = 119445.49 wich indicates a big spred of prices. Also we can see a very big range = 100000.0 that shows us a very big differnce between
# max and min price and could tell about some extremum prices in the listing. 
```

```python
variance_price = df['price'].var()
std_dev_price = df['price'].std()
min_price = df['price'].min()
max_price = df['price'].max()
range_price = max_price - min_price
q_price = df['price'].quantile([0.25, 0.5, 0.75])

print(variance_price)
print(std_dev_price)
print(range_price)
print(q_price)
```

```python
 #Visualizations**:
    # Is there a specific price range where most products fall? Plot a histogram to visualize the distribution of product prices. 
    # *If its hard to read these diagrams, think why this is, and explain how it could be solved.*.  

# Yes diagram is realy hard to read the reason is that we have a big spred of price. We can see this also on values of skewness_price and kurtosis_price. To ivode this we need to use
# data exluding outliers 
    # Are there products that are priced significantly higher than the rest? Use a box plot to showcase the spread and potential outliers in product pricing. 
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['price'], bins = 5, kde = True, color = 'blue' )

plt.title('Distribution of Product Prices')
plt.xlabel('Price')
plt.ylabel('Number of Products')
```

```python
skewness_price = df['price'].skew()
kurtosis_price = df['price'].kurtosis()

skewness_price, kurtosis_price
```
