```python
import pandas as pd

ruta = r"C:\Users\berta\OneDrive\Desktop\IRONHACK\SQL1\sakila\amz_uk_price_prediction_dataset.csv"

df = pd.read_csv(ruta)

df.head()
```

```python
df.columns
```

```python
category_counts = df["category"].value_counts()

category_counts
```

```python
top_5_categories = category_counts.head(5)

top_5_categories
```

```python
import matplotlib.pyplot as plt

top_5_categories.plot(kind="bar", figsize=(10, 5))

plt.title("Top 5 Product Categories on Amazon UK")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
top_5_categories.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(8, 8)
)

plt.title("Proportion of Listings Among the Top 5 Categories")
plt.ylabel("")
plt.show()
```

```python
mean_price = df["price"].mean()
median_price = df["price"].median()
mode_price = df["price"].mode()

print("Mean price:", mean_price)
print("Median price:", median_price)
print("Mode price:", mode_price.tolist())
```

```python
price_variance = df["price"].var()
price_std = df["price"].std()
price_range = df["price"].max() - df["price"].min()

q1 = df["price"].quantile(0.25)
q3 = df["price"].quantile(0.75)
iqr = q3 - q1

print("Variance:", price_variance)
print("Standard deviation:", price_std)
print("Range:", price_range)
print("Q1:", q1)
print("Q3:", q3)
print("Interquartile range (IQR):", iqr)
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["price"].dropna(), bins=50)

plt.title("Distribution of Amazon UK Product Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Products")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.boxplot(df["price"].dropna(), vert=False)

plt.title("Box Plot of Amazon UK Product Prices")
plt.xlabel("Price (£)")
plt.tight_layout()
plt.show()
```

```python
mean_stars = df["stars"].mean()
median_stars = df["stars"].median()
mode_stars = df["stars"].mode()

print("Mean rating:", mean_stars)
print("Median rating:", median_stars)
print("Mode rating:", mode_stars.tolist())
```

```python
stars_variance = df["stars"].var()
stars_std = df["stars"].std()

q1_stars = df["stars"].quantile(0.25)
q3_stars = df["stars"].quantile(0.75)
iqr_stars = q3_stars - q1_stars

print("Variance:", stars_variance)
print("Standard deviation:", stars_std)
print("Q1:", q1_stars)
print("Q3:", q3_stars)
print("IQR:", iqr_stars)
```

```python
print("Skewness:", df["stars"].skew())
print("Kurtosis:", df["stars"].kurt())
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df["stars"].dropna(), bins=20)
plt.title("Distribution of Product Ratings")
plt.xlabel("Stars")
plt.ylabel("Number of Products")
plt.tight_layout()
plt.show()
```
