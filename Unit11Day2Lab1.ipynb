### Part 2: Delving into Product Pricing

**Business Question**: How are products priced on Amazon UK, and are there specific price points or ranges that are more common?

1. **Measures of Centrality**:
    - Calculate the mean, median, and mode for the `price` of products.
    - What's the average price point of products listed? How does this compare with the most common price point (mode)?

2. **Measures of Dispersion**:
    - Determine the variance, standard deviation, range, and interquartile range for product `price`.
    - How varied are the product prices? Are there any indicators of a significant spread in prices?

3. **Visualizations**:
    - Is there a specific price range where most products fall? Plot a histogram to visualize the distribution of product prices. *If its hard to read these diagrams, think why this is, and explain how it could be solved.*.
    - Are there products that are priced significantly higher than the rest? Use a box plot to showcase the spread and potential outliers in product pricing. 


```python
# 1. Measures of Centrality: 
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

print("Measures of Centrality")
print(f"Mean Price: £{mean_price:.2f}")
print(f"Median Price: £{median_price:.2f}")
print(f"Mode Price: £{mode_price:.2f}")
```

```python
#2. Measures of Dispersion
variance_price = df['price'].var()
std_price = df['price'].std()
range_price = df['price'].max() - df['price'].min()
iqr_price = df['price'].quantile(0.75) - df['price'].quantile(0.25)

print("Measures of Dispersion")
print(f"Variance: {variance_price:.2f}")
print(f"Standard Deviation: {std_price:.2f}")
print(f"Range: {range_price:.2f}")
print(f"Interquartile Range (IQR): {iqr_price:.2f}")
```

```python
#3. Visualizations
# Histograms
plt.subplot(1, 2, 1)
sns.histplot(df['price'], bins=50, kde=False, color='skyblue')
plt.title('Distribution of Product Prices')
plt.xlabel('Price (£)')
plt.ylabel('Frequency')
```

```python
#Boxplot
plt.figure(figsize=(6, 8))
sns.boxplot(y=df['price'], color="lightgreen")

plt.title('Box Plot of Product Prices')
plt.xlabel('Price (£)')

plt.show()
```

Both charts are difficult to read due to the presence of extreme outliers (products with unusually high prices). These extreme values force the X-axis scale to expand, which compresses the distribution of the vast majority of products and hides the real trend

### Part 3: Unpacking Product Ratings

**Business Question**: How do customers rate products on Amazon UK, and are there any patterns or tendencies in the ratings?

1. **Measures of Centrality**:
    - Calculate the mean, median, and mode for the `rating` of products.
    - How do customers generally rate products? Is there a common trend?

2. **Measures of Dispersion**:
    - Determine the variance, standard deviation, and interquartile range for product `rating`.
    - Are the ratings consistent, or is there a wide variation in customer feedback?

3. **Shape of the Distribution**:
    - Calculate the skewness and kurtosis for the `rating` column. 
    - Are the ratings normally distributed, or do they lean towards higher or lower values?

4. **Visualizations**:
    - Plot a histogram to visualize the distribution of product ratings. Is there a specific rating that is more common?

---

```python
df_ratings = df.dropna(subset=['stars']).copy()
```

```python
zero_ratings = (df_ratings['stars'] == 0.0).sum()
zero_ratings
```

```python
df_clean_ratings = df_ratings[df_ratings['stars'] > 0].copy()
```

```python
# 1. Measures of Centrality: 
rating_mean = df_clean_ratings['stars'].mean()
rating_median = df_clean_ratings['stars'].median()
rating_mode = df_clean_ratings['stars'].mode()[0]

print("Corrected Measures of Centrality")
print(f"Mean Rating: {rating_mean:.2f}")
print(f"Median Rating: {rating_median:.2f}")
print(f"Mode Rating: {rating_mode:.2f}")
```

```python
#2. Measures of Dispersion: 
rating_variance = df_clean_ratings['stars'].var()
rating_std = df_clean_ratings['stars'].std()
rating_iqr = df_clean_ratings['stars'].quantile(0.75) - df_clean_ratings['stars'].quantile(0.25)

print("Measures of Dispersion (Clean Data)")
print(f"Variance: {rating_variance:.2f}")
print(f"Standard Deviation: {rating_std:.2f}")
print(f"Interquartile Range (IQR): {rating_iqr:.2f}")
```

```python
#3. Shape of the distribution
rating_skew = df_clean_ratings['stars'].skew()
rating_kurt = df_clean_ratings['stars'].kurt()

print("Shape of the Distribution")
print(f"Skewness: {rating_skew:.2f}")
print(f"Kurtosis: {rating_kurt:.2f}")
```

The product ratings are NOT normally distributed. 
The calculation shows a negative skewness, which indicates that the distribution leans heavily towards higher values (positive feedback), with a long tail stretching left towards the lower ratings. 
Furthermore, the positive kurtosis confirms that the distribution has a very sharp peak, meaning customer ratings are highly concentrated around the top scores rather than being evenly spread across the 1-to-5 scale.

```python
#4. Visualization
plt.figure(figsize=(10, 6))
sns.histplot(df_clean_ratings['stars'], bins=10, color='orange', kde=False)

plt.title('Distribution of Product Ratings on Amazon UK')
plt.xlabel('Rating (Stars)')
plt.ylabel('Frequency (Number of Products)')
plt.xticks([1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5])

plt.show()
```
