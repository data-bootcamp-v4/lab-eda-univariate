# Laboratorio: Análisis Univariado de Productos de Amazon UK

Este laboratorio realiza un análisis exploratorio univariado (EDA) de productos en Amazon Reino Unido para extraer insights de negocio sobre categorías, precios y calificaciones.

```
# Importar librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configuración de gráficos
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,6)

# Cargar dataset
data = pd.read_csv('ruta/del/archivo/amazon_products_uk.csv')

# Vista rápida del dataset
data.head()
```

## Parte 1: Comprensión de las categorías de productos

### Tabla de frecuencias de categorías

```
category_counts = data['category'].value_counts()
category_counts.head(10)
```

```
top5_categories = category_counts.head(5)
top5_categories
```

```
top10_categories = category_counts.head(10)
sns.barplot(x=top10_categories.values, y=top10_categories.index, palette="viridis")
plt.title("Top 10 categorías de productos en Amazon UK")
plt.xlabel("Número de productos")
plt.ylabel("Categoría")
plt.show()
```

```
top5_categories.plot(kind='pie', autopct='%1.1f%%', startangle=140, colors=sns.color_palette("viridis",5))
plt.ylabel('')
plt.title("Proporción de los 5 principales categorías de productos")
plt.show()
```

## Parte 2: Análisis de precios

```
mean_price = data['price'].mean()
median_price = data['price'].median()
mode_price = data['price'].mode()[0]
mean_price, median_price, mode_price
```

```
price_var = data['price'].var()
price_std = data['price'].std()
price_range = data['price'].max() - data['price'].min()
price_iqr = stats.iqr(data['price'])
price_var, price_std, price_range, price_iqr
```

```
sns.histplot(data['price'], bins=50, kde=True, color='skyblue')
plt.title("Distribución de precios de productos")
plt.xlabel("Precio")
plt.ylabel("Cantidad de productos")
plt.show()
```

```
sns.boxplot(x=data['price'], color='lightgreen')
plt.title("Diagrama de caja de precios de productos")
plt.xlabel("Precio")
plt.show()
```

## Parte 3: Análisis de calificaciones (ratings)

```
mean_rating = data['rating'].mean()
median_rating = data['rating'].median()
mode_rating = data['rating'].mode()[0]
mean_rating, median_rating, mode_rating
```

```
rating_var = data['rating'].var()
rating_std = data['rating'].std()
rating_iqr = stats.iqr(data['rating'])
rating_var, rating_std, rating_iqr
```

```
rating_skew = data['rating'].skew()
rating_kurt = data['rating'].kurt()
rating_skew, rating_kurt
```

```
sns.histplot(data['rating'], bins=10, kde=False, color='salmon')
plt.title("Distribución de calificaciones de productos")
plt.xlabel("Calificación")
plt.ylabel("Cantidad de productos")
plt.show()
```

## Conclusiones de negocio
- **Categorías**: Identificar las más populares permite priorizar marketing y gestión de inventario.
- **Precios**: Conocer rangos y detectar productos premium ayuda a ajustar pricing y promociones.
- **Ratings**: Analizar tendencias en calificaciones identifica productos bien valorados y posibles áreas de mejora.
