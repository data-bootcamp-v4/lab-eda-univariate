```python
import pandas as pd

df = pd.read_csv("amz_uk_price_prediction_dataset.csv")

print(df.shape)
df.head()

```

```python
category_freq = df["category"].value_counts()

print(category_freq)
```

```python
top5_categories = df["category"].value_counts().head(5)

print(top5_categories)
```

```python
plt.figure(figsize=(12,6))

top10_categories = df["category"].value_counts().head(10)

sns.barplot(
    x=top10_categories.values,
    y=top10_categories.index
)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Listings")
plt.ylabel("Category")
plt.show()
```

```python
top5_categories.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(8,8)
)

plt.title("Top 5 Product Categories")
plt.ylabel("")
plt.show()
```

```python
mean_price = df["price"].mean()
median_price = df["price"].median()
mode_price = df["price"].mode()[0]

print("Mean:", mean_price)
print("Median:", median_price)
print("Mode:", mode_price)
```

```python
variance_price = df["price"].var()
std_price = df["price"].std()

price_range = df["price"].max() - df["price"].min()

iqr_price = (
    df["price"].quantile(0.75)
    - df["price"].quantile(0.25)
)

print("Variance:", variance_price)
print("Std:", std_price)
print("Range:", price_range)
print("IQR:", iqr_price)
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["price"], bins=50)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df[df["price"] < df["price"].quantile(0.99)]["price"],
    bins=50
)

plt.title("Price Distribution (Without Extreme Outliers)")
plt.show()
```

```python
plt.figure(figsize=(10,2))

sns.boxplot(x=df["price"])

plt.title("Boxplot of Product Prices")
plt.show()
```

```python
mean_rating = df["stars"].mean()
median_rating = df["stars"].median()
mode_rating = df["stars"].mode()[0]

print("Mean:", mean_rating)
print("Median:", median_rating)
print("Mode:", mode_rating)
```

```python
variance_stars = df["stars"].var()
std_stars = df["stars"].std()

iqr_stars = (
    df["stars"].quantile(0.75)
    - df["stars"].quantile(0.25)
)

print("Variance:", variance_stars)
print("Standard Deviation:", std_stars)
print("IQR:", iqr_stars)
```

```python
skewness = df["stars"].skew()
kurtosis = df["stars"].kurt()

print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
```
