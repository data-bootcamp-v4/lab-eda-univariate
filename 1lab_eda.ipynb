```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Loading the housing price dataset
df = pd.read_csv('amz_uk_price_prediction_dataset.csv')
df
```

```python
# Generate frequency table for 'product_category'
freq_table = df['category'].value_counts()
freq_table
```

```python
# Select top 10 categories
top_categories = freq_table[:10]

# Plot bar chart
plt.figure(figsize=(12, 6))
sns.barplot(x=top_categories.values, y=top_categories.index, palette="viridis")

# Labels and title
plt.xlabel("Number of Products", fontsize=12)
plt.ylabel("Category", fontsize=12)
plt.title("Distribution of Products Across Categories", fontsize=14)
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for readability
plt.show()
```

```python

# Select top 5 product categories
top_categories = df['category'].value_counts().nlargest(5)

# Plot pie chart
plt.figure(figsize=(6, 6))
plt.pie(top_categories, labels=top_categories.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0'])

# Title
plt.title("Proportion of Top 5 Product Categories", fontsize=14)
plt.show()
```

```python

# Assuming 'price' column contains product prices
mean_price = df['price'].mean()  
median_price = df['price'].median()  
mode_price = df['price'].mode()[0]  

# Display results
print(f"Mean Price: £{mean_price:.2f}")
print(f"Median Price: £{median_price:.2f}")
print(f"Mode Price: £{mode_price:.2f}")
```

```python

plt.figure(figsize=(8,5))
sns.histplot(df['price'], bins=50, kde=True)
plt.xlabel('Product Price (£)')
plt.ylabel('Frequency')
plt.title('Distribution of Product Prices')
plt.show()
```
