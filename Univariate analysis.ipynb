```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%matplotlib inline
```

```python
# Define the URL of the CSV file
#url = 'https://www.kaggle.com/datasets/asaniczka/uk-optimal-product-price-prediction/'

# Read the CSV file from the URL
df = pd.read_csv("amz_uk_price_prediction_dataset.csv")

# Display the DataFrame
print(df)
df
```

```python
# Generate a frequency table for the product category
category_counts = df['category'].value_counts()

# Display the top 5 most listed product categories
top_5_categories = category_counts.head(5)
print("Top 5 Most Listed Product Categories:")
print(top_5_categories)
```

```python
# Sample data for the top 5 most listed product categories
data = {
    'category': [
        'Sports & Outdoors',
        'Beauty',
        'Handmade Clothing, Shoes & Accessories',
        'Bath & Body',
        'Birthday Gifts'
    ],
    'listing_frequency': [
        836265,
        19312,
        19229,
        19092,
        18978
    ]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Create a bar plot
plt.figure(figsize=(10, 6))
plt.bar(df['category'], df['listing_frequency'], color='skyblue')
plt.title('Top 5 Most Listed Product Categories on Amazon UK', fontsize=16)
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Number of Listings', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#2
import numpy as np
from scipy import stats

# Load the dataset (replace 'file_path' with the actual path to your dataset)
file_path = 'amz_uk_price_prediction_dataset.csv'  # Example file name
df = pd.read_csv(file_path)

# Calculate measures of centrality
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

print(f"Mean Price: {mean_price}")
print(f"Median Price: {median_price}")
print(f"Mode Price: {mode_price}")
```

```python
# Calculate measures of dispersion
variance_price = df['price'].var()
std_dev_price = df['price'].std()
price_range = df['price'].max() - df['price'].min()
q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)
iqr_price = q3 - q1

print(f"Variance: {variance_price}")
print(f"Standard Deviation: {std_dev_price}")
print(f"Range: {price_range}")
print(f"Interquartile Range (IQR): {iqr_price}")
```

```python
# Plot a histogram for product prices
plt.figure(figsize=(10, 6))
plt.hist(df['price'], bins=50, color='skyblue', edgecolor='black')
plt.title('Distribution of Product Prices', fontsize=16)
plt.xlabel('Price', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Plot a box plot for product prices
plt.figure(figsize=(8, 6))
plt.boxplot(df['price'], vert=False, patch_artist=True, boxprops=dict(facecolor='skyblue'))
plt.title('Box Plot of Product Prices', fontsize=16)
plt.xlabel('Price', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#3
import pandas as pd
import numpy as np
from scipy import stats

# Load the dataset (replace 'file_path' with the actual path to your dataset)
file_path = 'amz_uk_price_prediction_dataset.csv'  # Example file name
df = pd.read_csv(file_path)

# Calculate measures of centrality
mean_rating = df['stars'].mean()
median_rating = df['stars'].median()
mode_rating = df['stars'].mode()[0]

print(f"Mean Rating: {mean_rating}")
print(f"Median Rating: {median_rating}")
print(f"Mode Rating: {mode_rating}")
```

```python
# Calculate measures of dispersion
variance_rating = df['stars'].var()
std_dev_rating = df['stars'].std()
q1 = df['stars'].quantile(0.25)
q3 = df['stars'].quantile(0.75)
iqr_rating = q3 - q1

print(f"Variance: {variance_rating}")
print(f"Standard Deviation: {std_dev_rating}")
print(f"Interquartile Range (IQR): {iqr_rating}")
```

```python
from scipy.stats import skew, kurtosis

# Calculate skewness and kurtosis
skewness = skew(df['stars'])
kurt = kurtosis(df['stars'])

print(f"Skewness: {skewness}")
print(f"Kurtosis: {kurt}")
```

```python
import matplotlib.pyplot as plt

# Plot a histogram for product ratings
plt.figure(figsize=(10, 6))
plt.hist(df['stars'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Product Ratings', fontsize=16)
plt.xlabel('stars', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Plot a box plot for product ratings
plt.figure(figsize=(8, 6))
plt.boxplot(df['stars'], vert=False, patch_artist=True, boxprops=dict(facecolor='skyblue'))
plt.title('Box Plot of Product Ratings', fontsize=16)
plt.xlabel('stars', fontsize=12)
plt.tight_layout()
plt.show()
```
