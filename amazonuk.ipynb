```python
import pandas as pd

# Cargar el conjunto de datos
df = pd.read_csv('D:/Documents/Bootcamp/Amazon/amz_uk_price_prediction_dataset.csv')

# Generar tabla de frecuencia para 'category'
frecuencia_categoria = df['category'].value_counts()
print(frecuencia_categoria)
```

```python
# Verificar las primeras filas y las columnas del dataset
print(df.head())
print(df.info())
```

```python
# Tabla de frecuencias para las categorías de productos
category_counts = df['category'].value_counts()
print(category_counts)

# Top 5 categorías más listadas
top_5_categories = category_counts.head(5)
print(top_5_categories)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfico de barras para la distribución de categorías
plt.figure(figsize=(12, 6))
sns.barplot(x=top_5_categories.index, y=top_5_categories.values)
plt.xlabel('Categoría')
plt.ylabel('Número de Productos')
plt.title('Distribución de Productos por Categoría (Top 5)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Gráfico de pastel para las proporciones de las principales categorías
plt.figure(figsize=(8, 8))
plt.pie(top_5_categories, labels=top_5_categories.index, autopct='%1.1f%%', startangle=140)
plt.title('Proporción de Categorías de Productos (Top 5)')
plt.show()
```

```python
# Medidas de centralidad para el precio
mean_price = df['price'].mean()
median_price = df['price'].median()
mode_price = df['price'].mode()[0]

print(f'Precio Promedio: {mean_price}')
print(f'Precio Mediano: {median_price}')
print(f'Precio más Común: {mode_price}')
```

```python
# Medidas de dispersión para el precio
variance_price = df['price'].var()
std_dev_price = df['price'].std()
price_range = df['price'].max() - df['price'].min()
iqr_price = df['price'].quantile(0.75) - df['price'].quantile(0.25)

print(f'Varianza del Precio: {variance_price}')
print(f'Desviación Estándar del Precio: {std_dev_price}')
print(f'Rango del Precio: {price_range}')
print(f'Rango Intercuartil del Precio: {iqr_price}')
```

```python
# Histograma de precios
plt.figure(figsize=(12, 6))
sns.histplot(df['price'], bins=50, kde=True)
plt.xlabel('Precio')
plt.ylabel('Número de Productos')
plt.title('Distribución de Precios de Productos')
plt.show()
```

```python
# Box plot para detectar precios atípicos
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['price'])
plt.xlabel('Precio')
plt.title('Distribución de Precios de Productos con Box Plot')
plt.show()
```

```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv('D:/Documents/Bootcamp/Amazon/amz_uk_price_prediction_dataset.csv')

# Verificar los nombres de las columnas
print(df.columns)

# Verificar las primeras filas del DataFrame
print(df.head())

# Supongamos que el nombre correcto es 'rating'
if 'rating' in df.columns:
    # Medidas de centralidad para la calificación
    mean_rating = df['rating'].mean()
    median_rating = df['rating'].median()
    mode_rating = df['rating'].mode()[0]

    print(f'Calificación Promedio: {mean_rating}')
    print(f'Calificación Mediana: {median_rating}')
    print(f'Calificación más Común: {mode_rating}')

    # Medidas de dispersión para la calificación
    variance_rating = df['rating'].var()
    std_dev_rating = df['rating'].std()
    iqr_rating = df['rating'].quantile(0.75) - df['rating'].quantile(0.25)

    print(f'Varianza de la Calificación: {variance_rating}')
    print(f'Desviación Estándar de la Calificación: {std_dev_rating}')
    print(f'Rango Intercuartil de la Calificación: {iqr_rating}')

    # Cálculo de sesgo y curtosis
    from scipy.stats import skew, kurtosis

    skewness = skew(df['rating'].dropna())
    kurt = kurtosis(df['rating'].dropna())

    print(f'Sesgo de la Calificación: {skewness}')
    print(f'Curtosis de la Calificación: {kurt}')

    # Histograma de calificaciones
    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.figure(figsize=(12, 6))
    sns.histplot(df['rating'], bins=10, kde=True)
    plt.xlabel('Calificación')
    plt.ylabel('Número de Productos')
    plt.title('Distribución de Calificaciones de Productos')
    plt.show()
else:
    print("La columna 'rating' no se encuentra en el DataFrame.")
```
