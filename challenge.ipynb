```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
import seaborn as sns
import matplotlib.pyplot as plt

file_path = "amz_uk_price_prediction_dataset.csv"

df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "asaniczka/uk-optimal-product-price-prediction",
  file_path,
)

display(df)
```

```python
df.columns
```

```python
# Generate a frequency table for the product category.
unique_categories = df['category'].value_counts()
display(unique_categories) # Shows the full list of counts
```

```python
#Which are the top 5 most listed product categories?
unique_categories.head(5)
```

```python
unique_categories.head(10).plot(kind="bar",figsize=(10, 6), color='red')
plt.xlabel("Product Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
unique_categories.head(10).plot(kind="pie")
plt.title("Proportion of Top 10 Product Categories")
plt.show()
```

Yes Sports & Outdoors dominate the listings

```python
average_price = df['price'].mean()

mode_price = df['price'].mode()[0]

print(f"Average Price (Mean): {average_price:.2f}")
print(f"Most Common Price (Mode): {mode_price:.2f}")

difference = average_price - mode_price
print(f"\nThe average is {difference:.2f} higher than the most common price.")
```

```python

variance = df['price'].var()
std_deviation = df['price'].std()
price_range = df['price'].max() - df['price'].min()

q3 = df['price'].quantile(0.75)  
q1 = df['price'].quantile(0.25)
price_iqr = q3 - q1

print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {std_deviation:.2f}")
print(f"Range: {price_range:.2f}")
print(f"Interquartile Range (IQR): {price_iqr:.2f}")
```

The product prices are highly varied, but in a very specific way. A massive Range of 100,000.00 and a high Standard Deviation of 345.61 indicate a significant spread in prices

```python
variance = df['stars'].var()
std_deviation = df['stars'].std()
rating_range = df['stars'].max() - df['stars'].min()

q3 = df['stars'].quantile(0.75)  
q1 = df['stars'].quantile(0.25)
rating_iqr = q3 - q1

print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {std_deviation:.2f}")
print(f"Range: {rating_range:.2f}")
print(f"Interquartile Range (IQR): {rating_iqr:.2f}")
```

can see this because the Standard Deviation (2.19) is incredibly high for a 1-to-5 scale, and the Interquartile Range (4.40) spans nearly the entire rating spectrum. This wide variation exists because the data is polarized

```python
skew = df['stars'].skew()
kurtosis = df['stars'].kurtosis()

print(f"Skew: {skew:.2f}")
print(f"kurtosis: {kurtosis:.2f}")
```

The ratings are not normally distributed. A normal distribution would require most ratings to cluster in the middle (around 2.5 stars)

```python
import seaborn as sns
import matplotlib.pyplot as plt

filtered_rating = df[df['stars'] > 0]
sns.histplot(data=filtered_rating, x='stars', kde=True, bins=20, color='purple')

plt.title("Distribution of Actual Product Ratings (Excluding 0s)")
plt.xlabel("Rating (Stars)")
plt.ylabel("Frequency")
plt.show()
```
