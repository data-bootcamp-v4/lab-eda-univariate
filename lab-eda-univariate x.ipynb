```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
```

```python
df = pd.read_csv("amz_uk_price_prediction_dataset.csv")
```

```python
df.head()
```

```python
category_counts = df["category"].value_counts()

category_counts
```

```python
top5 = category_counts.head(5)

top5
```

```python
plt.figure(figsize=(12,6))

top5.plot(kind="bar")

plt.title("Top 5 Product Categories")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,8))

top5.plot(
    kind="pie",
    autopct="%1.1f%%",
    ylabel=""
)

plt.title("Top 5 Product Categories")

plt.show()
```

```python
df.columns.tolist()
```

### Business Insight

The distribution of product categories is uneven. The top five categories account for a large proportion of listings, suggesting that Amazon UK has a stronger presence in these segments. Businesses entering these categories may face higher competition, while less populated categories could represent niche opportunities.

```python
price = df["price"]

print("Mean:", price.mean())
print("Median:", price.median())
print("Mode:", price.mode()[0])
```

```python
print("Variance:", price.var())
print("Standard Deviation:", price.std())
print("Range:", price.max() - price.min())
print("Interquartile Range (IQR):", price.quantile(0.75) - price.quantile(0.25))
```

```python
plt.figure(figsize=(10,6))

plt.hist(price, bins=30)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,2))

plt.boxplot(price, vert=False)

plt.title("Boxplot of Product Prices")

plt.show()
```

### Business Insight

The average product price is influenced by a small number of very expensive products. The median provides a better representation of the typical product price. The histogram and boxplot reveal a right-skewed distribution with several high-price outliers.

```python
rating = df["stars"]

print("Mean:", rating.mean())
print("Median:", rating.median())
print("Mode:", rating.mode()[0])
```

```python
print("Variance:", rating.var())
print("Standard Deviation:", rating.std())
print("Interquartile Range (IQR):", rating.quantile(0.75) - rating.quantile(0.25))
```

```python
print("Skewness:", rating.skew())
print("Kurtosis:", rating.kurt())
```

```python
plt.figure(figsize=(10,6))

plt.hist(rating, bins=20)

plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

### Business Insight

Most products receive relatively high ratings, indicating generally positive customer satisfaction. The distribution is not perfectly normal and tends to be concentrated around higher values, suggesting that customers are more likely to leave positive reviews than negative ones.

# Final Conclusions

- The product category distribution is highly concentrated, with a small number of categories containing most listings.
- Product prices show a wide spread and are positively skewed due to expensive outliers.
- Customer ratings are generally high, indicating overall positive customer satisfaction.
- Median values provide a better measure of central tendency for prices because of the influence of extreme values.
- The visualizations reveal important patterns that can help businesses understand market competition, pricing strategies, and customer perception.
