```python
sns.histplot(df['price'], kde=True, bins=30, color="salmon")
plt.show()
```

```python
#A few extreme high prices compress most of the data into a few bars.the data is right skewed. (mean>median)
#Problem: Many real-world datasets (e.g., incomes, prices, ratings) have a long tail of high values.
#Solution: Logs shrink extreme values, making the distribution more normal (Gaussian).
```

```python
#log transformation
plt.figure(figsize=(10, 6))
sns.histplot(np.log10(df['price'] + 1), kde=True, bins=30, color="salmon")  # +1 to handle $0 prices
plt.title("Log10(Price) Distribution")
plt.xlabel("Log10(Price)")
plt.xticks(ticks=[0, 1, 2, 3, 4, 5], labels=['$1', '$10', '$100', '$1K', '$10K', '$100K'])
plt.show()
```

```python
bins = [0, 100, 500, 1000, 5000, 10000, 50000, 100000]
labels = ['0-100', '100-500', '500-1K', '1K-5K', '5K-10K', '10K-50K', '50K-100K']

# Create a new column for price categories
df['price_category'] = pd.cut(df['price'], bins=bins, labels=labels, include_lowest=True)
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='price_category', order=labels, palette='viridis')
plt.title('Number of Products by Price Category')
plt.xlabel('Price Range ($)')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#ignoring extreme outliers
plt.figure(figsize=(10, 6))
sns.histplot(df[df['price'] <= 1000]['price'], kde=True, bins=30, color="salmon")  # Adjust threshold (e.g., 1000)
plt.title("Price Distribution (≤ $1,000)")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
#most products are fall around the price range of between 10 and 55 dollars.
```

```python
#price data has extreme values (min=0, max=100,000), a standard boxplot is squished.#log scaled box-plot.
#Log transformation compresses high values more than low values (e.g., log10(1000) = 3, log10(10) = 1).
#The line in the log-scaled boxplot represents the median of log10(price), not the original price.
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=np.log10(df['price'] + 1),  # +1 handles $0 prices
            color="lightblue",
            width=0.3)
plt.title("Log10(Price) Distribution")
plt.xlabel("Log10(Price)")
plt.xticks([])  # Remove x-axis ticks (log scale is self-explanatory)
plt.gca().set_yticklabels(['$1', '$10', '$100', '$1K', '$10K', '$100K'])  # Optional: relabel y-axis
plt.show()
```

```python
df['stars'] = df['stars'].replace(0, pd.NA) #I replaced the zero values with NA.
mean_price=df["stars"].mean()
median_price =df["stars"].median()
mode_price=df["stars"].mode()[0]
mean_price, median_price, mode_price
```

```python
#The mean after excluding the zero ratings gives 4.31 which is less than the median-> Left skewed (negative skewed).
```

```python
sns.histplot(df[df['stars'] >= 1]['stars'], kde=True, bins=30, color="salmon")
plt.show()
```

```python
variance_stars = df['stars'].var()
std_dev_stars = df['stars'].std()
min_stars = df['stars'].min()
max_stars = df['stars'].max()
range_stars = max_stars - min_stars
quantiles_stars = df['stars'].quantile([0.25, 0.5, 0.75])
variance_stars, std_dev_price, min_stars, max_stars, range_stars, quantiles_stars
```

```python
#moderate standard deviation: 1, not a wide variation.
#most of the ratings are centered between 4 and 4.5
```

```python
skewness_stars = df['stars'].dropna().skew()
kurtosis_stars = df['stars'].dropna().kurtosis()

skewness_stars, kurtosis_stars
```

```python
#the distribution is negatively skewed. Positive kurtosis. Tail on the left.
```
